```python
df.columns = [
    'ER',
    'PR',
    'HER2',
    'Tumor_Size',
    'Nodes',
    'Stage'
]

df.head()
```

```python
def clean_status(x):
    if pd.isna(x):
        return None
    x = str(x).lower()

    if "pos" in x or "+" in x:
        return "Positive"
    elif "neg" in x or "-" in x:
        return "Negative"
    else:
        return None

df['ER'] = df['ER'].apply(clean_status)
df['PR'] = df['PR'].apply(clean_status)
df['HER2'] = df['HER2'].apply(clean_status)

df.head()
```

```python
df['Tumor_Size'] = pd.to_numeric(df['Tumor_Size'], errors='coerce')
df['Nodes'] = pd.to_numeric(df['Nodes'], errors='coerce')

df.head()
```

```python
def clean_stage(x):
    if pd.isna(x):
        return None
    x = str(x).upper()

    if "I" in x:
        return "I"
    elif "II" in x:
        return "II"
    elif "III" in x:
        return "III"
    elif "IV" in x:
        return "IV"
    else:
        return None

df['Stage'] = df['Stage'].apply(clean_stage)

df.head()
```

```python
df = df.dropna(subset=['ER', 'HER2', 'Nodes'])

df.shape
```

```python
df['Tumor_Size'] = df['Tumor_Size'] / 10
df.head()
```

```python
def assign_stage(row):
    if row['Nodes'] == 0 and row['Tumor_Size'] <= 2:
        return "I"
    elif row['Nodes'] <= 3:
        return "II"
    elif row['Nodes'] > 3:
        return "III"
    else:
        return None

df['Stage'] = df.apply(assign_stage, axis=1)

df.head()
```

```python
from transformers import AutoTokenizer, AutoModel

MODEL_NAME = "emilyalsentzer/Bio_ClinicalBERT"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME)

def get_embedding(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)
    outputs = model(**inputs)
    return outputs.last_hidden_state.mean(dim=1).detach().numpy()
```

```python
def clean_text(text):
    text = re.sub(r'-\s+', '', text)  # fix broken words
    text = re.sub(r'[^\w\s.,%()-]', '', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

```python
def extract_entities(sentence):
    sentence = sentence.lower()

    entities = {}

    if "her2 positive" in sentence:
        entities["HER2"] = "positive"
    elif "her2 negative" in sentence:
        entities["HER2"] = "negative"

    if "er positive" in sentence:
        entities["ER"] = "positive"
    elif "er negative" in sentence:
        entities["ER"] = "negative"

    if "node positive" in sentence:
        entities["Nodes"] = "positive"
    elif "node negative" in sentence:
        entities["Nodes"] = "negative"

    if "ki67" in sentence:
        if "high" in sentence:
            entities["Ki67"] = "high"
        elif "low" in sentence:
            entities["Ki67"] = "low"

    return entities
```

```python
def extract_treatment(sentence):
    sentence = sentence.lower()

    if "chemotherapy" in sentence:
        return "Chemotherapy recommended"
    elif "endocrine" in sentence or "tamoxifen" in sentence:
        return "Endocrine therapy recommended"
    elif "trastuzumab" in sentence or "her2-targeted" in sentence:
        return "HER2-targeted therapy recommended"

    return None
```

```python
COMORBIDITIES = ["diabetes", "hypertension", "renal failure", "cardiac disease", "pregnancy"]

def extract_comorbidities(sentence):
    found = []
    for c in COMORBIDITIES:
        if c in sentence.lower():
            found.append(c)
    return found
```

```python
def recommend_treatment(patient):

    # PRIORITY 1 — HER2
    if patient.get("HER2") == "positive":
        return {
            "recommendation": "HER2-targeted therapy + chemotherapy",
            "reason": "HER2 positive cancers benefit from targeted therapy",
            "confidence": "high"
        }

    # PRIORITY 2 — ER
    if patient.get("ER") == "positive":
        return {
            "recommendation": "Endocrine therapy",
            "reason": "Hormone receptor positive tumors respond to endocrine therapy",
            "confidence": "high"
        }

    # PRIORITY 3 — Nodes
    if patient.get("Nodes") == "positive":
        return {
            "recommendation": "Chemotherapy",
            "reason": "Node positive indicates higher spread risk",
            "confidence": "medium"
        }

    return {
        "recommendation": "Further evaluation needed",
        "reason": "Insufficient data",
        "confidence": "low"
    }
```

```python
def adjust_for_comorbidities(recommendation, comorbidities):

    if "cardiac disease" in comorbidities:
        if "HER2-targeted" in recommendation:
            return recommendation + " (Use cautiously due to cardiac risk)"

    if "pregnancy" in comorbidities:
        if "chemotherapy" in recommendation.lower():
            return "Avoid chemotherapy during pregnancy"

    return recommendation
```

```python
import json
from pprint import pprint

# ---- LOAD RULES ----
with open("rules.json", "r") as f:
    rules = json.load(f)

print(f"✅ Total Rules Loaded: {len(rules)}\n")


# ---- SHOW SAMPLE RULES ----
print("🔹 SAMPLE RULES:\n")
for i, r in enumerate(rules[:5]):
    print(f"Rule {i+1}:")
    pprint({
        "condition": r.get("condition"),
        "outcome": r.get("outcome"),
        "explanation": r.get("explanation")[:120] + "..."
    })
    print("-" * 50)


# ---- TEST PATIENT INPUT ----
patient = {
    "ER": "positive",
    "HER2": "negative",
    "Nodes": "negative"
}

print("\n🧍 TEST PATIENT:")
pprint(patient)


# ---- MATCH RULES ----
def match_rules(patient, rules):
    matched = []
    for rule in rules:
        condition = rule.get("condition", {})
        if all(patient.get(k) == v for k, v in condition.items()):
            matched.append(rule)
    return matched


matched = match_rules(patient, rules)

print(f"\n🎯 MATCHED RULES: {len(matched)}\n")

for i, r in enumerate(matched[:5]):
    print(f"Match {i+1}:")
    pprint({
        "condition": r.get("condition"),
        "outcome": r.get("outcome"),
        "explanation": r.get("explanation")[:120] + "..."
    })
    print("-" * 50)


# ---- PICK BEST RULE ----
def get_best_rule(matched_rules):
    if not matched_rules:
        return None
    matched_rules.sort(key=lambda x: len(x.get("condition", {})), reverse=True)
    return matched_rules[0]


best_rule = get_best_rule(matched)

print("\n🏆 BEST RULE:\n")

if best_rule:
    pprint({
        "condition": best_rule.get("condition"),
        "recommendation": best_rule.get("outcome"),
        "explanation": best_rule.get("explanation")
    })
else:
    print("❌ No matching rule found")
```
